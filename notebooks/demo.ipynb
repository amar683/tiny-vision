# tiny-vision demo

**Let your AI assistant watch videos.** tiny-vision is a 36M-parameter captioner (frozen
DINOv2-small encoder + a small caption decoder trained on COCO) that turns a video into a
scene-by-scene text breakdown any LLM can read.

Run the two cells below (**Runtime → Run all**). A demo app opens: upload a photo or a short
video (up to 3 minutes), then paste the breakdown into any LLM and ask *"what happens in this video?"*

No GPU needed. Captions are short and approximate (COCO style: no text reading, names or
counting), and it's visual only: no audio yet.

[GitHub](https://github.com/amar683/tiny-vision) · [Model](https://huggingface.co/amartomar683/tiny-vision)

### Or use it from Python directly

```python
from caption import load_run, caption_images
from vlm.scenes import describe_scenes
import torch
model, tok, cfg = load_run("amartomar683/tiny-vision", torch.device("cpu"))
text, _ = describe_scenes("my_video.mp4", lambda ims: caption_images(model, tok, cfg, ims, torch.device("cpu"), num_beams=3))
print(text)
```
